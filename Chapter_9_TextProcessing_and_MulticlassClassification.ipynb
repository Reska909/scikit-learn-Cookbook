## 1. Introduction
Feature extraction from text poses unique structural challenges. This chapter demonstrates how to clean text, perform vectorization (converting text into numerical matrix representations), and feed them into a multiclass classification model.

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report

# Synthetic text dataset: 3 categories (Technology, Sports, Food)
corpus = [
    "Apple rilis smartphone baru dengan prosesor AI canggih",
    "Google mengupdate algoritma machine learning di cloud",
    "Timnas menang dramatis lewat gol menit akhir di final piala AFF",
    "Pemain basket itu mencetak 50 poin dalam satu pertandingan",
    "Resep membuat nasi goreng jawa yang enak dan lezat",
    "Restoran ramen baru ini menyajikan kaldu ayam yang gurih"
]
y_labels = ['Technology', 'Technology', 'Sports', 'Sports', 'Food', 'Food']

print("Original Text Dataset:")
for i, text in enumerate(corpus):
    print(f"- {y_labels[i]}: '{text}'")
```

## 2. Text Vectorization (TF-IDF)
We utilize TF-IDF for text vectorization. It goes beyond simple word frequency matching by systematically down-weighting ubiquitous tokens that appear too frequently across all documents in the corpus.

```python
vectorizer = TfidfVectorizer()
X_tfidf = vectorizer.fit_transform(corpus)

print(f"TF-IDF Matrix Shape: {X_tfidf.shape} (Contains {X_tfidf.shape[1]} unique features/words)")
print("Sample extracted feature words:", vectorizer.get_feature_names_out()[:10])
```

## 3. Multiclass Classification with Pipelines
We employ the `MultinomialNB` variant of the Naive Bayes algorithm, which is highly optimized for discrete frequency data like text features. To keep our workflow modular and clean, we wrap both feature extraction and model prediction into a single `Pipeline`.

```python
# Construct an automated text-to-prediction Pipeline
model_pipeline = make_pipeline(TfidfVectorizer(), MultinomialNB())

# Fit the pipeline model
model_pipeline.fit(corpus, y_labels)

# Evaluation dataset (Unseen by the model during training)
test_docs = [
    "Cara memasak mie ayam super pedas",
    "Microsoft mengembangkan software AI untuk laptop",
    "Pelatih sepakbola tersebut merubah taktik babak kedua"
]
y_true = ['Food', 'Technology', 'Sports']

# Evaluate predictions
y_pred = model_pipeline.predict(test_docs)

print("=== Unseen Data Inference Results ===")
for doc, pred in zip(test_docs, y_pred):
    print(f"Text: '{doc}' \n--> Prediction: {pred}\n")

print("Classification Report:")
print(classification_report(y_true, y_pred))
```
